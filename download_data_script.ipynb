```python
# Requires: pip install astroquery

from astroquery.mast import MastMissions
from datetime import datetime, timezone
from pathlib import Path

missions = MastMissions(mission='JWST')
datasets = ['jw01286008001_31201_00001']
selected_keys = ['jw01286008001_31201_00001_jw01286008001_31201_00001_nrcb4_cal.fits']

# Retrieve products for the selected datasets, then filter down to exact selections
products = missions.get_product_list(datasets)
filtered = missions.filter_products(products, product_key=selected_keys)

# Prepare download directory (timestamped parent folder)
download_dir = Path("../data/")
download_dir.mkdir(exist_ok=True)
result = missions.download_products(filtered, download_dir=str(download_dir), flat=True)
print(f"Downloaded {len(result)} files to {download_dir}")
```
